```python
import pandas as pd
```

```python
volumen = pd.read_clipboard()
```

```python
indicadores = pd.read_clipboard()
```

```python
volumen.head()
```

```python
indicadores.head()
```

```python
(
    volumen
    .merge(
        indicadores,
        on=["indicador", "year"],
        how="inner"
    )
    .pivot(
        index=["carretera", "movimiento", "caseta", "kilometro"],
        columns="year",
        values="indicador"
    )
    # Nos quedamos solo con los radares permanentes que hayan estado durante
    # todo el tiempo que se esté estudiando los datos
    .pipe(lambda df: df[df.isna().astype(int).sum(axis=1) == 0])
    .melt(ignore_index=False, value_name='indicador')
    .reset_index()
)
```
